```python
v_cols = [c for c in train_tr.columns if c.startswith('V')]

corr = train_tr[v_cols].corr().abs()
upper = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
highly_corr = [col for col in upper.columns if any(upper[col] > 0.95)]
perfect_corr = [col for col in upper.columns if any(upper[col] == 1)]

print(f'{len(v_cols)} total V columns in train')
print(f'{len(highly_corr)} are >0.95 correlated with another V column')
print(f'{len(perfect_corr)} are perfectly correlated with another V column')
```

```python
v_cols = [c for c in test_tr.columns if c.startswith('V')]

corr = test_tr[v_cols].corr().abs()
upper = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
highly_corr = [col for col in upper.columns if any(upper[col] > 0.95)]
perfect_corr = [col for col in upper.columns if any(upper[col] == 1)]

print(f'{len(v_cols)} total V columns in test')
print(f'{len(highly_corr)} are >0.95 correlated with another V column')
print(f'{len(perfect_corr)} are perfectly correlated with another V column')
```

## Notes:

*   20,663 cases of fraud positively flagged in train dataset (~3.5% of total)
    *   Might make assessing fraud detection accuracy a bad way to go for assessing model performance
    *   Maybe ROC-AUC or PR-AUC instead so we can catch false positives?
    *   May be good to consider adding some class weights when training models
*   Time gap between train and test: train runs day 0 to ~day 182, test runs ~day 213 to ~day 396, so there's about a 30-day gap between where train ends and test starts
    *   May cause trouble if we want to combine and re-split it on our own
          * Also would need to figure out how to create "isFraud" flag for new rows from test set, currently doesn't exist in test data
*   A lot of V columns have same missing pct rate; likely means if one col has a value present, it will also be present in many other V cols (haven't tested to confirm this 100% yet)
    *   High correlation between V cols
         *   Train: 124 cols >95% correlated to other V cols, 0 perfectly correlated
         *   Test: 84 cols >95% correlated, 1 perfect
    *  Wouldn't be the best to impute vals for this since over 80% of the vals are null
         *   How do we want to treat those null values? most models don't handle nulls very well
         *   Gradient boosted model uses nulls as a meaningful indicator instead of as a missing val (i think), maybe we should look into that?
    *   Maybe figure out if we can drop some of these cols for training to reduce noise
*   Want to look deeper into what unique values in each col are missing in train vs. in test:
    *   'card5' (119 train vs 102 test)
    *   'card6' (4 unique in train vs 3 in test)
    *   If they are rare cases, maybe lump into "other" col for easier processing?

*   Some cols that do show a significant difference in the amount of fraud detected b/t unique values:
    *   ProductCD, C 11.7% vs W 2.0%, big spread
    *   card4 and card6
    *   card2, card3, card5 may be too noisy, group or bin these?
    *   P_emaildomain
    *   Didn't check all so potentially others

## To-dos for model training pre-processing:
* Logistic regression:
    *   No missing values unless we find a library that can handle it. maybe use "missing" flag somehow instead of NaN/null?
    *   Regularization between numerical values
         *   TransactionAmt contains vals in the thousands (max is over 31k), operates on a different scale than other cols that may have a smaller range. need to get all columns operating on the same scale to make them more comparable or need to figure out how to make sure model is processing them correctly

# Identity Datasets

```python
train_id = pd.read_csv('train_identity.csv')
test_id = pd.read_csv('test_identity.csv')

print(f'train_identity: {train_id.shape}   test_identity: {test_id.shape}')
```

```python
train_id.head()
```

```python
test_id.head()
```

```python
# fix col names
test_id.columns = [c.replace('-', '_') for c in test_id.columns]
test_id.head()
```

```python
train_tr_merge = train_tr[['TransactionID', 'isFraud']]
match_rate = train_tr_merge['TransactionID'].isin(train_id['TransactionID']).mean()
print(f'% of train transactions that have an identity record: {match_rate * 100:.2f}%')
```

```python
merged = train_tr_merge.merge(train_id[['TransactionID']], on='TransactionID', how='left', indicator=True)
print('fraud rate, has identity record vs. does not')
print(train_tr_merge.assign(has_identity=train_tr_merge['TransactionID'].isin(train_id['TransactionID']))
      .groupby('has_identity')['isFraud'].agg(['count', 'mean']))
```

```python
print('train')
print(train_id.dtypes.value_counts())

print('\ntest')
print(test_id.dtypes.value_counts())
```

```python
print('sorted missing vals train')
miss_train = train_id.isnull().mean().sort_values(ascending=False)
print((miss_train[miss_train > 0] * 100).round(2))

print('\nsorted missing vals train')
miss_test = test_id.isnull().mean().sort_values(ascending=False)
print((miss_test[miss_test > 0] * 100).round(2))
```

```python
# based on metadata
cat_cols = ['DeviceType', 'DeviceInfo'] + [f'id_{i}' for i in range(12, 39)]
cat_cols = [c for c in cat_cols if c in train_id.columns]

print('train set unique vals per column')
for col in cat_cols:
    print(f'{col}: {train_id[col].nunique(dropna=True)}')
```

```python
print('test set unique vals per column')
for col in cat_cols:
    print(f'{col}: {test_id[col].nunique(dropna=True)}')
```

```python
num_cols = [c for c in train_id.columns if c.startswith('id_') and c not in cat_cols]
print('numeric ID cols describe (train)')
print(train_id[num_cols].describe().T)
```

```python
num_cols = [c for c in test_id.columns if c.startswith('id_') and c not in cat_cols]
print('numeric ID cols describe (test)')
print(test_id[num_cols].describe().T)
```

```python
print('DeviceType value counts (train)')
print(train_id['DeviceType'].value_counts(dropna=False))

print('\nDeviceInfo top 20 value counts (train)')
print(train_id['DeviceInfo'].value_counts(dropna=False).head(20))
```

```python
print('DeviceType value counts (test)')
print(test_id['DeviceType'].value_counts(dropna=False))

print('\nDeviceInfo top 20 value counts (test)')
print(test_id['DeviceInfo'].value_counts(dropna=False).head(20))
```

```python
merged_full = train_id.merge(train_tr_merge, on='TransactionID', how='left')

print('fraud rate by DeviceType')
print(merged_full.groupby('DeviceType', dropna=False)['isFraud'].agg(['count', 'mean']))

print('\nfraud rate by top 10 DeviceInfo')
top_devices = merged_full['DeviceInfo'].value_counts().head(10).index
print(merged_full[merged_full['DeviceInfo'].isin(top_devices)]
      .groupby('DeviceInfo')['isFraud'].agg(['count', 'mean'])
      .sort_values('count', ascending=False))
```

```python
print(f'in train not test: {set(train_id.columns) - set(test_id.columns)}')
print(f'in test not train: {set(test_id.columns) - set(train_id.columns)}')

for col in ['DeviceInfo', 'id_30', 'id_31', 'id_33']:
    if col in train_id.columns and col in test_id.columns:
        new_vals = set(test_id[col].dropna().unique()) - set(train_id[col].dropna().unique())
        total = test_id[col].nunique(dropna=True)
        pct = 100 * len(new_vals) / total if total else 0
        print(f'{col}: {len(new_vals)} values in test never seen in train ({pct:.1f}% of test\'s unique values)')
```
