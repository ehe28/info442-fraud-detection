```python
identity = pd.read_csv('/content/train_identity.csv')

# if a transactionID is in both datasets, flag it
df_final['hasID'] = df_final['TransactionID'].astype(str).isin( identity['TransactionID'].astype(str)).astype(int)

identity = None
```

## Dealing with Categorical Values

```python
# create list of categorical values
cat_features = df.select_dtypes(include=['object']).columns
# bucket certain columns into groups
for col in cat_features:
  display(df[col].value_counts())
```

```python
# make T/F columns binary
t_f = ['M2', 'M1', 'M3', 'M5', 'M6', 'M7', 'M8', 'M9']

for col in t_f:
  df_final.loc[df_final[col] == 'T', col] = 1
  df_final.loc[df_final[col] == 'F', col] = 0
  df_final.loc[df_final[col].isna(), col] = -999 # null imputation
  df_final[col] = df_final[col].astype(int)

df_final[t_f].value_counts()
```

```python
df_final[t_f].head()
```

```python
# lump debit or credit and charge card into other category
df_final.loc[df['card6'] == 'debit or credit', 'card6'] = 'Other'
df_final.loc[df['card6'] == 'charge card', 'card6'] = 'Other'

df_final['card6'].value_counts()
```

```python
# transform DX columns to be constant for each UID despite delta value
# we can determine what is important during feature selection.
date_col = ['D1', 'D2', 'D3', 'D4', 'D5', 'D6', 'D7', 'D8', 'D9', 'D10', 'D11','D12','D13','D14','D15']
for col in date_col:
  df_final[col] = df_final['TransactionDT'] - df_final[col]
```
