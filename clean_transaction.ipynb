```python
# load data, libraries
import pandas as pd
import numpy as np

df = pd.read_csv('/content/train_transaction.csv')

# join test and train data

df.head()
```

```python
identity = pd.read_csv('/content/train_identity.csv')

df['TransactionID_str'] = df['TransactionID'].astype(str)
identity['TransactionID'] = identity['TransactionID'].astype(str)

# if a transactionID is in both datasets, flag it
df['hasID'] = df['TransactionID_str'].isin(identity['TransactionID']).astype(int)
```

```python
# transform DX columns to be constant for each UID despite delta value
# we can determine what is important during feature selection.

df['D1'] = df['TransactionDT'] - df['D1'] # only value important for fining UIDs, almost everything else will likely be a subset of this
df['D2'] = df['TransactionDT'] - df['D2']
df['D3'] = df['TransactionDT'] - df['D3']
df['D4'] = df['TransactionDT'] - df['D4']
df['D5'] = df['TransactionDT'] - df['D5']
df['D6'] = df['TransactionDT'] - df['D6']
df['D7'] = df['TransactionDT'] - df['D7']
df['D8'] = df['TransactionDT'] - df['D8']
df['D9'] = df['TransactionDT'] - df['D9']
df['D10'] = df['TransactionDT'] - df['D10']
df['D11'] = df['TransactionDT'] - df['D11']
df['D12'] = df['TransactionDT'] - df['D12']
df['D13'] = df['TransactionDT'] - df['D13']
df['D14'] = df['TransactionDT'] - df['D14']
df['D15'] = df['TransactionDT'] - df['D15']
```

```python
df[['D1', 'D2', 'D3', 'D4', 'D5', 'D6', 'D7', 'D8', 'D9', 'D10', 'D11','D12','D13','D14','D15']].fillna(-999, inplace=True)
```
